```python
#Task 07: String Validation Methods

text = "NLP is the future of AI research"

if text.startswith("NLP"):
    print("The text starts with NLP")
else:
    print("The text does not start with NLP")

if text.endswith("research"):
    print("The text ends with research")
else:
    print("The text does not end with research")

print("Position of future:", text.find("future"))


tokens = ["hello", "123", "world", "456abc", "python", "789", "NLP3"]

for token in tokens:
    if token.isalpha():
        print(token, "- Alphabetic")
    elif token.isdigit():
        print(token, "- Numeric")
    else:
        print(token, "- Mixed")


words = ["running", "jumped", "NLP", "ai2025", "data", "42"]

pure_words = []
pure_numbers = []
mixed_tokens = []

for word in words:
    if word.isalpha():
        pure_words.append(word)
    elif word.isdigit():
        pure_numbers.append(word)
    else:
        mixed_tokens.append(word)

print("Pure words:", pure_words)
print("Pure numbers:", pure_numbers)
print("Mixed tokens:", mixed_tokens)


def validate_token(token):
    if token.isalpha():
        return "WORD"
    elif token.isdigit():
        return "NUMBER"
    else:
        return "MIXED"


test_tokens = ["chat", "99", "gpt4", "bert", "2024", "nlp!"]

for token in test_tokens:
    print(token, "->", validate_token(token))
```

TASK 8

```python
#Task 08: Functions and Lambda for NLP Pipelines

def clean(text):
    text = text.strip()
    text = text.lower()
    return text

clean_lambda = lambda text: text.strip().lower()

text = "   HELLO Natural Language Processing   "

print(clean(text))
print(clean_lambda(text))


def tokenize(text):
    return text.split()


def remove_short(tokens, n):
    result = []
    for token in tokens:
        if len(token) > n:
            result.append(token)
    return result


sentence = "I love working with NLP tools in Python daily"

tokens = tokenize(sentence)
tokens = remove_short(tokens, 3)

print(tokens)


def word_lengths(tokens):
    lengths = {}
    for word in tokens:
        lengths[word] = len(word)
    return lengths


tokens = ["language", "model", "ai", "processing", "deep"]

print(word_lengths(tokens))


def nlp_pipeline(text, min_len):
    text = clean(text)
    tokens = tokenize(text)
    tokens = remove_short(tokens, min_len)
    return word_lengths(tokens)


sentence1 = "   Natural Language Processing is very interesting   "
sentence2 = "   Deep learning helps machines understand text   "

print(nlp_pipeline(sentence1, 3))
print(nlp_pipeline(sentence2, 4))
```

```python
#Task 09: map(), filter(), sorted() and Counter

words = ["Python", "NLP", "is", "a", "powerful", "tool", "for", "text"]

lowercase_words = list(map(lambda word: word.lower(), words))
print(lowercase_words)

word_lengths = list(map(lambda word: len(word), words))
print(word_lengths)


words_long = list(filter(lambda word: len(word) > 3, words))
words_with_t = list(filter(lambda word: "t" in word.lower(), words))

print(words_long)
print(words_with_t)


alphabetical = sorted(words)
by_length = sorted(words, key=len)
by_length_descending = sorted(words, key=len, reverse=True)

print("Alphabetically:", alphabetical)
print("By length:", by_length)
print("By length descending:", by_length_descending)


from collections import Counter

sentence = "nlp nlp python ai nlp python text ai nlp python text"

sentence_words = sentence.split()
freq = Counter(sentence_words)

print(freq)
print(freq.most_common(3))
print("Python frequency:", freq["python"])
print("Total words:", len(sentence_words))
print("Unique vocabulary:", len(freq))
```
